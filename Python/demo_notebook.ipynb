```r
# Imports
!pip install numpy
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print('✅ Tous les imports ont réussi!')
print(f'NumPy version: {np.__version__}')
print(f'Pandas version: {pd.__version__}')
```

```r
# Créer des données
data = pd.DataFrame({
    'x': np.arange(0, 10, 0.1),
    'y': np.sin(np.arange(0, 10, 0.1))
})

data.head()
```

```r
# Visualisation
plt.figure(figsize=(10, 6))
plt.plot(data['x'], data['y'], linewidth=2)
plt.title('Fonction Sinus', fontsize=16)
plt.xlabel('X')
plt.ylabel('Y')
plt.grid(True, alpha=0.3)
plt.show()
```
