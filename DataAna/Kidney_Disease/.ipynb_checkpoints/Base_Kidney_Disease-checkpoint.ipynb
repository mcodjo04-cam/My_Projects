```python
# Colonnes numériques qui contiennent des valeurs non numériques (dates, ?, espaces)
cols_to_convert_to_numeric = ['sg', 'bu', 'sc', 'sod', 'pot', 'hemo', 'pcv', 'wc', 'rc'] 
print("debut de la conversion des colonnes 'object en 'float' ")
for col in cols_to_convert_to_numeric:
    # pd.to_numeric avec errors='coerce' transforme les dates/chaînes non numériques en NaN
    data[col] = pd.to_numeric(data[col], errors='coerce') 
print("la conversion est termine")    
```

```python
# Liste de TOUTES les colonnes numériques
numerical_cols = ['age', 'bp', 'sg', 'al', 'su', 'bgr', 'bu', 'sc', 'sod', 'pot','hemo', 'pcv', 'wc', 'rc']
print("debut de l'inputation")
for col in numerical_cols:
    col_median = data[col].median()
    #NaN remplace par la mediane
    data[col].fillna(col_median, inplace=True)
print("fin!")    
    #Pourquoi la médiane ? Elle est choisie car, contrairement à la moyenne, elle n'est pas sensible aux valeurs aberrantes (outliers) extrêmes (comme des taux de créatinine très élevés), ce qui est crucial dans les données médicales.#
```

```python
# Colonnes catégorielles
categorical_cols = ['rbc', 'pc', 'pcc', 'ba', 'htn', 'dm', 'cad', 'appet', 'pe', 'ane', 'classification']

print("on commence le netoyage des variables categorielles")

for col in categorical_cols:
    data[col] = data[col].astype(str).str.lower().str.strip()
    
    # Remplacement des valeurs non valides par NaN (si cela n'a pas été fait)
    data[col].replace(to_replace=['?', '', '\t', 'nan'], value=np.nan, inplace=True)
    
    # 2. Imputation des NaN par la categorie missing
    data[col].fillna('missing', inplace=True)

print("On a fini le nettoyage et l'imputation des catégorielles terminée.")
```

## Encodage Binaire avec Missing

```python
# 0 pour (no,notpresent,normal,good) , 1(yes,present,abnormal,poor,) et 2 pour(missing)
print("on commence le mappage")
binary_mapping = {
    'yes': 1, 'no': 0,
    'present': 1, 'notpresent': 0,
    'abnormal': 1, 'normal': 0,
    'poor': 1, 'good': 0,
    'missing': 2 
}
#on veut encoder la variable cible classification
#ckd:1,notckd:0 et missng:0
data['classification'].replace({'ckd': 1, 'notckd': 0, 'missing': 0}, inplace=True) 

cols_to_map = [col for col in categorical_cols if col != 'classification']

for col in cols_to_map:
    data[col].replace(binary_mapping, inplace=True)
    data[col] = data[col].astype(float)
print("on a fini!")
```

## Detection des outliers

```python
def find_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    # Filtrer pour trouver les outliers
    outliers = data[(data[column] < lower_bound) | (data[column] > upper_bound)]
    
    # Afficher le nombre d'outliers et les seuils
    print(f"--- {column} ---")
    print(f"Outliers trouvés: {len(outliers)}")
    print(f"Seuil Inférieur: {lower_bound:.2f}, Seuil Supérieur: {upper_bound:.2f}")
    
    return outliers

# Essayons avec la Créatinine Sérique (sc)
outliers_sc = find_outliers_iqr(data, 'pot')
```
