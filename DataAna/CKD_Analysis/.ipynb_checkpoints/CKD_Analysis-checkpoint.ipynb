### Librairies utilisees

```python
#
import numpy as numpy
import pandas as pd 
#
import matplotlib.pyplot as plt 
import seaborn as sns 
#
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
from catboost import CatBoostRegressor
```

```python
### Importation des donnees
df = pd.read_csv('kidney_disease.csv')
df.head(10)
```

#### Renommer les colonnes

1-il n'est pas facile de comprendre les variables, il doivent etre renommer

```python
new_col= ['id','age','pression_arterielle','gravite_specifique','albumine','sucre','globule','cellules_pus','agregation_pus','bacteries','glucose_sanguin','uree','creatinine','soduim','potassium','hemoglobine','volume_cellules','globules_blanc','globules_rouges','hypertension','diabete','maladie_coronarienne','appetit','oedeme_pieds','anemie','classification']

df.columns = new_col

df.head(200)
```

```python
df.info()
```

2-On remarque qu'il a des valeurs qui manquent qui doivent etre geres.

3-les types des variables doivent etre revues

#### Nettoyage et conversion des colonnes numeriques mal types

```python
# Liste des colonnes numériques actuellement mal typées (object)
cols_num = ['volume_cellules', 'globules_blanc', 'globules_rouges']

for col in cols_num:
    # 'errors='coerce'' il convertit les chaînes invalides en NaN
    df[col] = pd.to_numeric(df[col], errors='coerce')

    #convertir en entier si possible, en gérant les NaN (utilisant 'Int64')
    df[col] = df[col].astype('Int64', errors='ignore') 
```

####  Conversion des variables categorielles

```python
# Liste des colonnes catégorielles
cols_categoriel = [
    'globule', 'cellules_pus', 'agregation_pus', 'bacteries', 
    'hypertension', 'diabete', 'maladie_coronarienne', 
    'appetit', 'oedeme_pieds', 'anemie', 'classification'
]

# Convertir toutes ces colonnes en type 'category'
for col in cols_categoriel:
    df[col] = df[col].astype('category')
```

### Imputation des donnees 

#### sur les colonnes numeriques

```python
# Crée une copie propre (deep copy) du DataFrame avec les types corrigés
df_nettoye = df.copy()
```

```python
# Identifier les colonnes numériques dans la copie de travail
cols_num_a_imputer = df_nettoye.select_dtypes(include=['float64', 'Int64']).columns.tolist()

# Exclure l'ID si elle est présente (elle n'a pas besoin d'être imputée)
if 'id' in cols_num_a_imputer:
    cols_num_a_imputer.remove('id')
    
print(f"Colonnes numériques à imputer : {cols_num_a_imputer}")
```

```python
# 1. Définir le modèle de régression CatBoost
cbr = CatBoostRegressor(verbose=0, # Entraînement silencieux
                        loss_function='RMSE', # Mieux pour la régression
                        random_state=42)

# 2. Configurer l'IterativeImputer
imputer_num = IterativeImputer(estimator=cbr, 
                               max_iter=10, 
                               random_state=42,
                               # S'assurer que le résultat est dans le bon format (array)
                               imputation_order='random') 


df_nettoye[cols_num_a_imputer] = imputer_num.fit_transform(df_nettoye[cols_num_a_imputer])

print("\nImputation des variables numériques terminée.")
```

```python
# Liste des colonnes qui doivent être des entiers (avec support NaN)
cols_int64_corrige = [
    'age', 
    'albumine', 
    'sucre', 
    'volume_cellules', 
    'globules_blanc'
]

for col in cols_int64:
    if col in df_nettoye.columns:
       df_nettoye[col] = df_nettoye[col].round().astype('Int64')
    
print("Types Int64 restaurés.")
```

```python
### colonnes categorielles
# Identifier les colonnes catégorielles avec NaN
cols_cat = df_nettoye.select_dtypes(include=['category']).columns.tolist()
# Filtrer celles qui ont encore des NaN
cols_cat_a_imputer = [col for col in cols_cat if df_nettoye[col].isnull().any()]

for col in cols_cat_a_imputer:
    # Calculer le mode (la valeur la plus fréquente)
    mode_val = df_nettoye[col].mode()[0] 
    
    # Remplacer les NaN
    df_nettoye[col].fillna(mode_val, inplace=True)
    
print("Imputation des variables catégorielles par le Mode terminée.")
```

```python
valeurs_manquantes_finales = df_nettoye.isnull().sum().sum()

if valeurs_manquantes_finales == 0:
    print(f"\n Nettoyage des données terminé : {valeurs_manquantes_finales} valeurs manquantes restantes.")
else:
    print(f"\n ERREUR : Il reste {valeurs_manquantes_finales} valeurs manquantes. Veuillez vérifier le DataFrame.")
```
