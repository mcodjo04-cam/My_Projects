```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
df = pd.read_csv("ventes.xlsx - Sheet1.csv")
```

```python
df.head()
```

```python
df.info()
```

#### nous remarquons qu'il y a des valeurs manquante au niveau de prix_unitaire et quantite

```python
df.describe()
```

```python
comptage_produits = df['produit'].value_counts()
print("\n Nombre de ventes par produit :")
print(comptage_produits)

# Trouver le produit le plus vendu
produit_star = comptage_produits.index[0]
print(f"\n Produit le PLUS vendu : {produit_star} ({comptage_produits.iloc[0]} ventes)")

# GRAPHIQUE 1 : Barres pour les produits
plt.figure(figsize=(10, 6))
comptage_produits.plot(kind='bar', color='skyblue', edgecolor='black')
plt.title(' Nombre de ventes par produit', fontsize=16, fontweight='bold')
plt.xlabel('Produit', fontsize=12)
plt.ylabel('Nombre de ventes', fontsize=12)
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('1_produits_univarie.png', dpi=300, bbox_inches='tight')
print("Graphique sauvegardé : 1_produits_univarie.png")
plt.close()
```

```python
comptage_vendeurs = df['vendeur'].value_counts()
print("\n Nombre de ventes par vendeur :")
print(comptage_vendeurs)

meilleur_vendeur = comptage_vendeurs.index[0]
print(f"\nMeilleur vendeur : {meilleur_vendeur} ({comptage_vendeurs.iloc[0]} ventes)")

# GRAPHIQUE 2 : Camembert pour les vendeurs
plt.figure(figsize=(8, 8))
plt.pie(comptage_vendeurs, labels=comptage_vendeurs.index, autopct='%1.1f%%', 
        startangle=90, colors=plt.cm.Set3.colors)
plt.title('Répartition des ventes par vendeur', fontsize=16, fontweight='bold')
plt.savefig('2_vendeurs_univarie.png', dpi=300, bbox_inches='tight')
print("Graphique sauvegardé : 2_vendeurs_univarie.png")
plt.close()
```

```python
comptage_regions = df['région'].value_counts()
print("\n Nombre de ventes par région :")
print(comptage_regions)

# GRAPHIQUE 3 : Barres horizontales pour les régions
plt.figure(figsize=(8, 6))
comptage_regions.plot(kind='barh', color='lightgreen', edgecolor='black')
plt.title('Ventes par région', fontsize=16, fontweight='bold')
plt.xlabel('Nombre de ventes', fontsize=12)
plt.ylabel('Région', fontsize=12)
plt.tight_layout()
plt.savefig('3_regions_univarie.png', dpi=300, bbox_inches='tight')
print("Graphique sauvegardé : 3_regions_univarie.png")
plt.close()
```

```python
print("\n Statistiques des prix unitaires :")
print(f"  • Prix minimum    : {df['prix_unitaire'].min():,.0f} FCFA")
print(f"  • Prix maximum    : {df['prix_unitaire'].max():,.0f} FCFA")
print(f"  • Prix moyen      : {df['prix_unitaire'].mean():,.0f} FCFA")
print(f"  • Prix médian     : {df['prix_unitaire'].median():,.0f} FCFA")

# GRAPHIQUE 4 : Histogramme des prix
plt.figure(figsize=(10, 6))
plt.hist(df['prix_unitaire'], bins=20, color='coral', edgecolor='black')
plt.title(' Distribution des prix', fontsize=16, fontweight='bold')
plt.xlabel('Prix (FCFA)', fontsize=12)
plt.ylabel('Fréquence', fontsize=12)
plt.axvline(df['prix_unitaire'].mean(), color='red', linestyle='--', linewidth=2, label='Prix moyen')
plt.legend()
plt.tight_layout()
plt.savefig('4_prix_univarie.png', dpi=300, bbox_inches='tight')
print(" Graphique sauvegardé : 4_prix_univarie.png")
plt.close()
```

```python
print("\n Statistiques des quantités vendues :")
print(f"  • Quantité minimum : {df['quantité'].min()}")
print(f"  • Quantité maximum : {df['quantité'].max()}")
print(f"  • Quantité moyenne : {df['quantité'].mean():.2f}")

# GRAPHIQUE 5 : Boîte à moustaches
plt.figure(figsize=(8, 6))
plt.boxplot(df['quantité'], vert=True, patch_artist=True,
            boxprops=dict(facecolor='lightblue'))
plt.title(' Distribution des quantités', fontsize=16, fontweight='bold')
plt.ylabel('Quantité', fontsize=12)
plt.tight_layout()
plt.savefig('5_quantites_univarie.png', dpi=300, bbox_inches='tight')
print(" Graphique sauvegardé : 5_quantites_univarie.png")
plt.close()
```

#### Analyse bivarié

```python
# Tableau croisé : compter les ventes par produit et région
tableau_prod_region = pd.crosstab(df['produit'], df['région'])
print("\n📊 Tableau : Produits × Régions")
print(tableau_prod_region)

# GRAPHIQUE 6 : Graphique à barres groupées
plt.figure(figsize=(12, 6))
tableau_prod_region.plot(kind='bar', figsize=(12, 6), colormap='Set2')
plt.title(' Ventes par Produit et Région', fontsize=16, fontweight='bold')
plt.xlabel('Produit', fontsize=12)
plt.ylabel('Nombre de ventes', fontsize=12)
plt.legend(title='Région', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.savefig('6_produit_region_bivarie.png', dpi=300, bbox_inches='tight')
print(" Graphique sauvegardé : 6_produit_region_bivarie.png")
plt.close()
```

```python
tableau_vendeur_produit = pd.crosstab(df['vendeur'], df['produit'])
print("\n📊 Tableau : Vendeurs × Produits")
print(tableau_vendeur_produit)

# GRAPHIQUE 7 : Heatmap (carte de chaleur)
plt.figure(figsize=(14, 6))
sns.heatmap(tableau_vendeur_produit, annot=True, fmt='d', cmap='YlOrRd', 
            linewidths=0.5, cbar_kws={'label': 'Nombre de ventes'})
plt.title('Carte de chaleur : Vendeur × Produit', fontsize=16, fontweight='bold')
plt.xlabel('Produit', fontsize=12)
plt.ylabel('Vendeur', fontsize=12)
plt.tight_layout()
plt.savefig('7_vendeur_produit_bivarie.png', dpi=300, bbox_inches='tight')
print(" Graphique sauvegardé : 7_vendeur_produit_bivarie.png")
plt.close()
```
